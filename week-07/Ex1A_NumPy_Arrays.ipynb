```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])

print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through each row in the array, so it multiplies the whole row by 2.
The second loop goes inside each row and prints each individual number multiplied by 2.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because loop_result uses x * 2, not y * 2.
# x represents the whole row, so it prints the whole row multiplied by 2.
```

```python
## Joining and Stacking Arrays
```

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# Error explanation:
# This gives an error because joined_array_1 and joined_array_2 have different shapes.
# NumPy cannot stack arrays unless they have the same shape.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# Error explanation:
# This gives an error because the arrays do not have the same number of rows.
# joined_array_1 has 4 rows, but arrA has 2 rows
```

```python
h_stack = np.hstack((arrA, arrA))  # hstack joins arrays horizontally, side by side
v_stack = np.vstack((arrA, arrA))  # vstack joins arrays vertically, top to bottom
d_stack = np.dstack((arrA, arrA))  # dstack stacks arrays by depth, creating a 3D array

print("hstack:")
print(h_stack)

print("vstack:")
print(v_stack)

print("dstack:")
print(d_stack)
```

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)

# This splits joined_array_1 into 2 arrays.
# Each array is 2-dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# This splits joined_array_1 into 3 arrays.
# Each array is still 2-dimensional, but the rows are not split evenly
```

```python
## Searching and Replacing Array Values
```

```python
arrB = np.array([
    [7, 6],
    [13, 5],
    [7, 3]
])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7
arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7
arrD = arrB[y]

print(arrD)
```
