## Permutations

This section demonstrates shuffle and permutation operations.

```python
print("Original r_4:")
print(r_4)
```

```python
np.random.shuffle(r_4)

print("Shuffled r_4:")
print(r_4)
```

The shuffle function changes the original array directly by rearranging the rows.
The values are mixed in a random order.

```python
new_shuffle = np.random.permutation(r_4)

print("Permutation Result:")
print(new_shuffle)

print("\nOriginal r_4:")
print(r_4)
```

Shuffle changes the original array itself, while permutation creates a new shuffled copy.
Use shuffle when you want to permanently rearrange data.
Use permutation when you want to keep the original array unchanged.

## Teamwork

This section tracks door-knocking activity for a voter registration campaign.

```python
ali = np.array([5, 4, 6, 4, 2])

ben = np.array([0, 5, 5, 3, 4])

carla = np.array([3, 6, 0, 3, 4])

print("Ali:", ali)
print("Ben:", ben)
print("Carla:", carla)
```

```python
all_knocks = np.array([ali, ben, carla])

print(all_knocks)
```

```python
person_totals = np.sum(all_knocks, axis=1)

group_total = np.sum(all_knocks)

print("Total doors knocked by each person:")
print(person_totals)

print("\nGroup Total:")
print(group_total)
```

```python
daily_totals = np.sum(all_knocks, axis=0)

rolling_total = np.cumsum(daily_totals)

estimated_votes = rolling_total * 0.5

print("Estimated Rolling Votes:")
print(estimated_votes)
```
