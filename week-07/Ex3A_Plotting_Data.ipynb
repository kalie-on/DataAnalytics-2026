# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
print(x)
print(y)
```

```python
plt.plot(x,y)
plt.show
```

```python
plt.plot(x, y, marker='o')

plt.show()
```

```python
plt.plot(x, y, marker='o')

plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

## Styling Plots

```python
plt.plot(x, y, linestyle='--', color='red')

plt.title("Dashed Red Line")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

```python
plt.plot(
    x,
    y,
    linestyle='--',
    color='red',
    marker='s',
    markersize=10,
    markerfacecolor='green'
)

plt.title("Square Green Markers")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

```python
plt.plot(
    x,
    y,
    marker='^',
    color='blue'
)

plt.title("Triangle Markers")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

```python
plt.plot(
    x,
    y,
    marker='*',
    color='purple'
)

plt.title("Star Markers")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
# I am creating product and sales data for my bar chart
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
```

```python
# I am creating a bar chart to compare sales by product

plt.bar(products, sales)

plt.show()
        
```

```python
# I am customizing my bar chart with a little, labels, color and bar width

plt.bar(products, sales, color="skyblue", width=0.5)

plt.title("sales by product")
plt.xlabel("products")
plt.ylabel("sales")

plt.show()
```

```python
# I am creating random x and y values for any scatter plot
x = np.random.rand(50)
y = np.random.rand(50)
```

```python
# I am creating marker sizes for my scatter plot
sizes = np.random.rand(50) * 100

# I am creating a scatter plot to show the relationship between x and y
plt.scatter(x, y, s=sizes, c=sizes, cmap='viridis')

plt.title("My Random Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

## Subplots

```python
# I am creating data for my sine wave line chart
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
# I am creating 1000 random numbers for my histogram
random_numbers = np.random.randn(1000)
```

```python
# I am creating a figure with two subplots

fig, axes = plt.subplots(1,2, figsize=(12,5))

# first subplot; sine wave line chart
axes[0].plot(x, y)
axes[0].set_title("My since wave line chart")
axes[0].set_xlabel("x value")
axes[0].set_ylabel("sine values")

# second subplot: histogram
axes[1].hist(random_numbers)
axes[1].set_title("my histogrom of random numbers")
axes[1].set_xlabel("random values")
axes[1].set_ylabel("frequency")

plt.tight_layout()
plt.show()
```
