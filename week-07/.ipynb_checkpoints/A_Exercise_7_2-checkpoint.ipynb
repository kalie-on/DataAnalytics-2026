## 2. Inspect and Describe the Data

In this section, we inspect the first rows, last rows, and summary statistics of the dataset.

```python
print("First 3 rows:")
print(df.head(3))

print("\nLast 3 rows:")
print(df.tail(3))

print("\nSummary Statistics:")
print(df.describe())

print(f'\nMean sales: £{df["sales"].mean():,.2f}')
print(f'Max sales: £{df["sales"].max():,.2f}')
print(f'Min sales: £{df["sales"].min():,.2f}')
print(f'Total sales: £{df["sales"].sum():,.2f}')
```

## 3. Select, Filter, and Sort Data

This section demonstrates selecting columns, filtering rows, and sorting data.

```python
print("Name and Sales Columns:")
print(df[["name", "sales"]])

print("\nEmployees with sales greater than £3500:")
high_sales = df[df["sales"] > 3500.0]
print(high_sales)

df_sorted = df.sort_values("sales", ascending=False)

print("\nSorted by sales highest to lowest:")
print(df_sorted)

print("\nTop-selling employee:")
print(df_sorted.iloc[0])
```

## 4. Add a Calculated Column and Conditional Rating

We create a commission column and assign performance ratings using a lambda function.

```python
df["commission"] = df["sales"] * 0.08

df["rating"] = df["sales"].apply(
    lambda x: "Star" if x >= 4000
    else ("Good" if x >= 2500 else "Developing")
)

print("Updated DataFrame:")
print(df)

print("\nCommission per employee:")

for _, row in df.iterrows():
    print(f'{row["name"]}: £{row["commission"]:,.2f} | {row["rating"]}')
```

## 5. GroupBy — Aggregate Statistics by City

This section groups the data by city and calculates total and average sales.

```python
city_totals = df.groupby("city")["sales"].sum()

print("Total Sales by City:")
print(city_totals)

city_means = df.groupby("city")["sales"].mean()

print("\nMean Sales by City:")
print(city_means)

city_summary = df.groupby("city").agg({
    "sales": ["sum", "mean", "count"]
})

print("\nFull City Summary:")
print(city_summary)

top_city = city_totals.idxmax()

print(f"\nTop city: {top_city} Total: £{city_totals[top_city]:,.2f}")
```
