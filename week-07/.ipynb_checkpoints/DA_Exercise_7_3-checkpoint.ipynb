This notebook reads customer order and shipment CSV files into Pandas DataFrames.
The notebook cleans, merges, and analyzes the data to create a complete order fulfilment report for GlobalTrade Ltd.

## 1. Import Pandas and Check Files
This section imports pandas and verifies that the CSV files exist in the current folder

```python
import pandas as pd
import os

print(os.listdir('.'))
```

## 2. Load CSV Files
This section loads orders.csv and shipments.csv into Pandas DataFrames.

```python
df_orders = pd.read_csv('orders.csv')

df_shipments = pd.read_csv('shipments.csv')
```

## 3. Preview the Orders DataFrame
Display the first 5 rows, shape, data types, and missing values

```python
print(df_orders.head())

print("\nSHAPE")
print(df_orders.shape)

print("\nDTYPES")
print(df_orders.dtypes)

print("\nMISSING VALUES")
print(df_orders.isnull().sum())
```

## 4. Preview the Shipments DataFrame
Display the first 5 rows, shape, data types, and missing values.

```python
print(df_shipments.head())

print("\nSHAPE")
print(df_shipments.shape)

print("\nDTYPES")
print(df_shipments.dtypes)

print("\nMISSING VALUES")
print(df_shipments.isnull().sum())
```

## 5. Create order_value Column
This section calculates the total value of each order.

```python
df_orders['order_value'] = (
    df_orders['quantity'] *
    df_orders['unit_price']
)

print(df_orders)
```

## 6. Numeric Summary Statistics
Use describe() to display summary statistics.

```python
print(df_orders.describe())
```

## 7. Convert ship_date to Datetime
This converts the ship_date column from object/string into datetime format.

```python
df_shipments['ship_date'] = pd.to_datetime(
    df_shipments['ship_date']
)

print(df_shipments.dtypes)
```

## 8. Total Order Value
Calculate the total order value for all customer orders.

```python
total = df_orders['order_value'].sum()

print(f"Total order value: £{total:,.2f}")
```

## 9. INNER MERGE
Only matching order_id values from both DataFrames are kept.

```python
df_inner = pd.merge(
    df_orders,
    df_shipments,
    on='order_id',
    how='inner'
)

print(df_inner)

print("\nSHAPE")
print(df_inner.shape)
```

## 10. LEFT MERGE
Keep all orders even if shipment information is missing.

```python
df_left = pd.merge(
    df_orders,
    df_shipments,
    on='order_id',
    how='left'
)

print(df_left)
```

## 11. Find Unshipped Orders
Rows with missing carrier values are unshipped orders.

```python
unshipped = df_left[
    df_left['carrier'].isna()
]

print(unshipped)
```

## 12. RIGHT MERGE
Keep all shipment rows even if order information is missing.

```python
df_right = pd.merge(
    df_orders,
    df_shipments,
    on='order_id',
    how='right'
)

print(df_right)
```

## 13. Find Orphan Shipment
Find shipment rows without matching order records.

```python
orphan = df_right[
    df_right['customer'].isna()
]

print(orphan)
```

## 14. OUTER MERGE
Keep all rows from both DataFrames.

```python
df_outer = pd.merge(
    df_orders,
    df_shipments,
    on='order_id',
    how='outer'
)

print(df_outer)

print("\nSHAPE")
print(df_outer.shape)

print("\nTOTAL NaN VALUES")
print(df_outer.isnull().sum().sum())
```

## 15. Add Shipped Flag
Create a True/False column showing whether the order shipped.

```python
df_left['shipped'] = (
    df_left['carrier'].notna()
)

print(df_left[
    ['order_id',
     'customer',
     'order_value',
     'shipped']
])
```

## 16. Count Shipped vs Unshipped Orders

```python
print(
    df_left['shipped'].value_counts()
)
```

## 17. Fulfilment Rate
Calculate the percentage of shipped orders.

```python
shipped_count = df_left['shipped'].sum()

total_orders = len(df_left)

fulfilment_rate = (
    shipped_count /
    total_orders * 100
)

print(
    f"Fulfilment rate: {fulfilment_rate:.2f}%"
)
```

## 18. GroupBy Summary by Status
Analyze order values by shipment status

```python
status_summary = (
    df_inner.groupby('status')
    ['order_value']
    .agg(['sum', 'mean', 'count'])
)

print(status_summary)
```

## 19. Carrier Totals
Display total order values by carrier

```python
carrier_totals = (
    df_inner.groupby('carrier')
    ['order_value']
    .sum()
    .sort_values(ascending=False)
)

print(carrier_totals)

print(
    f"\nTop carrier: {carrier_totals.index[0]}"
)
```

## 20. Units Dispatched by Status
Display total quantity shipped for each status.

```python
units_per_status = (
    df_inner.groupby('status')
    ['quantity']
    .sum()
)

print(units_per_status)
```

- FedEx had the highest total shipment value.
- Order ID 111 is an orphan shipment because it does not have a matching order record.
