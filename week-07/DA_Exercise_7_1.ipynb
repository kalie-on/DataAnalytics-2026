This notebook uses NumPy to analyze weekly revenue data for five regional offices.  
It covers array creation, descriptive statistics, arithmetic operations, and 2-D reshaping

## 1. Import NumPy and Create the Revenue Array

```python
import numpy as np
```

```python
revenue = np.array([52_400.0, 38_750.0, 61_200.0, 47_900.0, 55_650.0])

print(f"Array : {revenue}")
print(f"dtype : {revenue.dtype}")
print(f"shape : {revenue.shape}")
```

## 2. Descriptive Statistics

```python
print(f"Total : ${np.sum(revenue):,.2f}")
print(f"Mean : ${np.mean(revenue):,.2f}")
print(f"Median : ${np.median(revenue):,.2f}")
print(f"Std Dev : ${np.std(revenue):,.2f}")
print(f"Minimum : ${np.min(revenue):,.2f}")
print(f"Maximum : ${np.max(revenue):,.2f}")
```

## 3. Element-Wise Arithmetic — 5% Bonus

```python
revenue_bonus = revenue * 1.05

bonus_amount = revenue_bonus - revenue

print("Original Revenue :", revenue)

print("Revenue + 5% Bonus :", revenue_bonus)

print("Bonus Amount :", bonus_amount)
```

## 4. Boolean Masking — Above-Average Regions

```python
avg = np.mean(revenue)

above_avg = revenue > avg

print(f"Mean : ${avg:,.2f}")

print("Mask (T/F) :", above_avg)

print("Above average :", revenue[above_avg])

print(f"Regions above : {np.sum(above_avg)} of {len(revenue)}")
```

## 5. Reshape and 2-D Analysis

```python
col_vec = revenue.reshape(5, 1)

print("Column vector (5,1):\n", col_vec)
```

```python
q_data = np.vstack([revenue, revenue * 1.05])

print("\n2-D array (2 weeks × 5 regions):\n", q_data)
```

```python
print("Weekly totals :", np.sum(q_data, axis=1))

print("Region avg (2wk):", np.mean(q_data, axis=0))
```

## Summary

- The five regions generated a combined total of $256,900 in Week 1.
- Region 3 ($61,200) is the top performer, while Region 2 ($38,750) is the lowest.
- 3 out of 5 regions exceeded the average revenue of $51,380.
- Applying the 5% bonus increased the Week 2 total revenue.
